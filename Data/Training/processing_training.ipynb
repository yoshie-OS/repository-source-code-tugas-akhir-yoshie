```python
import re
import pandas as pd

def clean_tweet(text):
    text = re.sub(r'\[USERNAME\]', '', text)
    text = re.sub(r'\[URL\]', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'&\w+;', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

train_df = pd.read_csv("training_dataset.csv")
train_df['tweet'] = train_df['tweet'].apply(clean_tweet)

# verifikasi
remaining = train_df['tweet'].str.contains(r'http|@|\[USERNAME\]|\[URL\]|&\w+;', na=False).sum()
print(f"Sisa yang belum bersih: {remaining}")
print(train_df['tweet'].head(5))

# train_df.to_csv("training_dataset_cleaned.csv", index=False)
print("Saved.")
```
